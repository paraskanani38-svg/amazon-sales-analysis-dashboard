```python
# data cleaning
```

```python
df[['discounted_price', 'actual_price']].head(5)
```

```python
df['discounted_price'] = df['discounted_price'].str.replace('₹', '')
df['discounted_price'] = df['discounted_price'].str.replace(',', '')

df['actual_price'] = df['actual_price'].str.replace('₹', '')
df['actual_price'] = df['actual_price'].str.replace(',', '')
```

```python
df['discounted_price'] = df['discounted_price'].astype(float)
df['actual_price'] = df['actual_price'].astype(float)
```

```python
print(df[['discounted_price', 'actual_price']].dtypes)
```

```python
df[['discounted_price', 'actual_price']].head()
```

```python
df['discount_percentage'] = df['discount_percentage'].str.replace('%', '')
df['discount_percentage'] = df['discount_percentage'].astype(float)
print(df['discount_percentage'].dtype)
df['discount_percentage'].head()
```

```python
df['rating_count'].head()
```

```python
df['rating_count'].isnull().sum()
df = df.dropna(subset=['rating_count'])
df['rating_count'] = df['rating_count'].str.replace(',', '')
df['rating_count'] = df['rating_count'].astype(float)
```

```python
df['category'].head()
```

```python
df['main_category'] = df['category'].str.split('|').str[0]
df[['category', 'main_category']].head()
```

```python
df['main_category'].unique()
```

```python
df['product_id'].duplicated().sum()
```

```python
df[df['product_id'].duplicated()]
df = df.drop_duplicates(subset='product_id')
df['product_id'].duplicated().sum()
```

```python
print(df.shape)
```

```python
df['savings'] = df['actual_price'] - df['discounted_price']
df[['actual_price', 'discounted_price', 'savings']].head()
```

```python
print(df['savings'].dtype)
```

```python
df['main_category'] = df['category'].str.split('|').str[0]
df['main_category'].head()
clean_df = df.copy()
```

```python
df['main_category'] = df['category'].str.split('|').str[0]
df['sub_category']  = df['category'].str.split('|').str[1]

print('main_category unique values:', df['main_category'].nunique())
print('sub_category  unique values:', df['sub_category'].nunique())
df[['category', 'main_category', 'sub_category']].head(10)
clean_df = df.copy()
```

```python
def rating_group(r):
    if r >= 4:
        return 'High (4-5 Star)'
    elif r >= 3:
        return 'Medium (3 Star)'
    else:
        return 'Low (Below 3 Star)'

df['rating'] = pd.to_numeric(df['rating'], errors='coerce')
df['rating_group'] = df['rating'].apply(rating_group)
df['rating_group'].value_counts()
clean_df = df.copy()
```

```python
bins   = [0, 20, 40, 60, 100]
labels = ['0-20%', '21-40%', '41-60%', '61%+']

df['discount_bucket'] = pd.cut(
    df['discount_percentage'],
    bins=bins,
    labels=labels,
    include_lowest=True
)
df['discount_bucket'].value_counts().sort_index()
clean_df = df.copy()
```

```python
top_10_categories = df['main_category'].value_counts().head(10)

plt.figure(figsize=(10, 6))
top_10_categories.sort_values().plot(kind='barh')

plt.title('Top 10 Product Categories')
plt.xlabel('Number of Products')
plt.ylabel('Category')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.hist(
    df['discount_percentage'].dropna(),
    bins=20
)

plt.title('Discount Distribution')
plt.xlabel('Discount Percentage')
plt.ylabel('Number of Products')

plt.tight_layout()
plt.show()
```

```python
df['rating'] = pd.to_numeric(df['rating'], errors='coerce')

plt.figure(figsize=(10,6))

sns.countplot(
    data=df,
    x='rating',
    order=sorted(df['rating'].dropna().unique())
)

plt.title('Rating Distribution')
plt.xlabel('Rating')
plt.ylabel('Number of Products')

plt.show()
```

```python
plt.figure(figsize=(10,6))

plt.scatter(
    df['actual_price'],
    df['discounted_price'],
    alpha=0.5
)

plt.title('Actual Price vs Discounted Price')
plt.xlabel('Actual Price')
plt.ylabel('Discounted Price')

plt.show()
```

```python
top_reviewed = (
    df[['product_name', 'rating_count']]
    .dropna()
    .sort_values('rating_count', ascending=False)
    .head(10)
)

print(top_reviewed)
```

```python
top_reviewed = top_reviewed.sort_values('rating_count')

plt.figure(figsize=(12,6))

plt.barh(
    top_reviewed['product_name'],
    top_reviewed['rating_count']
)

plt.title('Top 10 Most Reviewed Products')
plt.xlabel('Number of Reviews')
plt.ylabel('Product Name')

plt.tight_layout()
plt.show()
```

```python
df['discount_category'] = pd.cut(
    df['discount_percentage'],
    bins=[0, 25, 50, 75, 100],
    labels=['0-25%', '26-50%', '51-75%', '76-100%']
)

discount_breakdown = df['discount_category'].value_counts()

plt.figure(figsize=(8,8))
plt.pie(
    discount_breakdown,
    labels=discount_breakdown.index,
    autopct='%1.1f%%'
)
plt.title('Discount Category Breakdown')
plt.show()
```

```python
df['rating'] = pd.to_numeric(df['rating'], errors='coerce')

df['main_category'] = df['category'].str.split('|').str[0]

avg_rating = df.groupby('main_category')['rating'].mean().sort_values(ascending=False)

plt.figure(figsize=(12,6))
avg_rating.plot(kind='bar')

plt.title('Average Rating by Main Category')
plt.xlabel('Main Category')
plt.ylabel('Average Rating')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df['actual_price'] = pd.to_numeric(df['actual_price'], errors='coerce')

df['main_category'] = df['category'].str.split('|').str[0]

top_categories = df['main_category'].value_counts().head(10).index

filtered_df = df[df['main_category'].isin(top_categories)]

boxplot_data = [
    filtered_df[filtered_df['main_category'] == cat]['actual_price'].dropna()
    for cat in top_categories
]

plt.figure(figsize=(14, 7))
plt.boxplot(boxplot_data, labels=top_categories, patch_artist=True)

plt.title('Price Range Distribution by Category')
plt.xlabel('Main Category')
plt.ylabel('Actual Price (₹)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
